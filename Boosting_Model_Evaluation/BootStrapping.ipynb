# Bootstrapping: Intuitive Notes

## 1. What is Bootstrapping?
- Bootstrapping is a statistical resampling technique.
- **Core idea:** From a dataset of size $n$, create new datasets (of size $n$) by sampling **with replacement**.
- The same data point can appear multiple times in a new dataset, while some points might not appear at all.
- Each new dataset is called a **bootstrap sample**.

## 2. Why Do We Do This?
- In real life, we usually have only one dataset.
- Bootstrapping simulates having many datasets.
- Helps estimate uncertainty (variance, confidence intervals) of our model.

## 3. Example (Step by Step)
Suppose we have dataset: $D = [2, 4, 6, 8, 10]$
- **Step 1: Bootstrap sample**
  - Randomly pick 5 numbers with replacement:
    - $D_1 = [4, 6, 6, 10, 2]$
    - $D_2 = [2, 2, 4, 8, 10]$
    - $D_3 = [6, 8, 8, 2, 4]$
  - Notice: some numbers repeat, some disappear.
- **Step 2: Train models (if used in ML)**
  - Train one model on $D$, and others on $D_1$, $D_2$, $D_3$, ...
  - Each model is slightly different because the training sets are different.

## 4. The Probability Part
- **What is the chance a data point is not selected in a bootstrap sample?**
  - Each draw: probability of not picking that point = $(1 - \frac{1}{n})$.
  - For $n$ draws: $P(\text{point not chosen}) = (1 - \frac{1}{n})^n$
  - As $n \to \infty$, this approaches $\frac{1}{e} \approx 0.368$
- So, in each bootstrap sample:
  - About 63.2% of the original data appears.
  - About 36.8% is left out (**out-of-bag data**).
- Out-of-bag points can be used to validate the model.

## 5. Bootstrapping in Bagging
- **Bagging** uses bootstrap:
  - Each tree is trained on a slightly different version of the dataset.
  - Trees become different, but still learn the same task.
  - Their averaged output is more stable.

## 6. Real-Life Analogy
- Imagine judging a singing contest 🎤 with 100 contestants (original dataset).
- Different groups of judges sample contestants randomly (with replacement) and give scores.
- Some singers get judged multiple times, some not at all.
- At the end, you average all scores → more reliable.
- That’s bootstrapping!

## ✅ Recap in Simple Words
- Bootstrapping = resampling with replacement.
- Each bootstrap sample ≈ ~63% of original data (some duplicates).
- Used in ML to create multiple datasets → train multiple models → reduce variance.
