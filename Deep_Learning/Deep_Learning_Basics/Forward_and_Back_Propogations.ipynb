# Forward and Backpropagation in Neural Networks

**Forward propagation** = calculating the output.  
**Backpropagation** = correcting the weights when the output is wrong.

Let's build this step by step, with math and a simple example.

---

## 1️⃣ Forward Propagation (Prediction Phase)

Think of a neural network as a big function with knobs (weights).

For a 1 hidden layer network:

- **Input:** $x$
- **Hidden layer:** $h = f(Wx + b)$
- **Output:** $\hat{y} = g(Vh + c)$

Where:
- $W, V$ = weights (knobs we tune)
- $b, c$ = biases
- $f, g$ = activation functions (like ReLU, sigmoid, etc.)
- $\hat{y}$ = predicted output

So the forward pass is just plugging in inputs, multiplying by weights, applying activations, and predicting.

### 🔢 Example

Say we want to predict whether a student will pass (1) or fail (0) an exam.

- Hours studied: $x_1 = 5$
- Sleep hours: $x_2 = 7$

Weights (random initial):

$$
W = \begin{bmatrix} 0.2 & 0.4 \\ 0.3 & 0.1 \end{bmatrix}, \quad b = \begin{bmatrix} 0.5 \\ 0.2 \end{bmatrix}
$$

Hidden layer calculation:

$$
Wx = \begin{bmatrix} 0.2 & 0.4 \\ 0.3 & 0.1 \end{bmatrix} \begin{bmatrix} 5 \\ 7 \end{bmatrix} = \begin{bmatrix} (0.2\times5 + 0.4\times7) \\ (0.3\times5 + 0.1\times7) \end{bmatrix} = \begin{bmatrix} 3.8 \\ 2.2 \end{bmatrix}
$$

Add bias:

$$
\begin{bmatrix} 3.8+0.5 \\ 2.2+0.2 \end{bmatrix} = \begin{bmatrix} 4.3 \\ 2.4 \end{bmatrix}
$$

Apply ReLU (max(0, x)):

$$
h = \begin{bmatrix} 4.3 \\ 2.4 \end{bmatrix}
$$

Now output layer:

- $V = [0.6, 0.9]$, $c = 0.1$

$$
z = Vh + c = (0.6\times4.3 + 0.9\times2.4) + 0.1 = 5.29
$$

Apply sigmoid (for binary classification):

$$
\hat{y} = \sigma(5.29) \approx 0.995
$$

**Prediction:** Student will pass (1) with high probability.

---

## 2️⃣ Backpropagation (Learning Phase)

Now suppose the true label is $y = 1$.

We compute loss (error). For binary classification, cross-entropy loss is common:

$$
L = -[y \log(\hat{y}) + (1-y)\log(1-\hat{y})]
$$

If prediction is off, backprop works like this:

- Compute error at output (difference between $\hat{y}$ and $y$):

$$
\delta_{\text{output}} = \hat{y} - y
$$

- Send this error backwards through the network using chain rule of calculus.
- For hidden neurons:

$$
\delta_h = (V^T \delta_{\text{output}}) \cdot f'(z_h)
$$

- Update weights using gradient descent:

$$
W = W - \eta \frac{\partial L}{\partial W}, \quad V = V - \eta \frac{\partial L}{\partial V}
$$

where $\eta$ = learning rate.

### 🔢 Example (continuing above)

- Predicted $\hat{y} = 0.995$
- Actual $y = 1$

Output error:

$$
\delta_{\text{output}} = 0.995 - 1 = -0.005
$$

Backpropagate into hidden layer (for first hidden neuron):

$$
\delta_{h1} = V_1 \cdot \delta_{\text{output}} \cdot f'(z_{h1})
$$

Update weights (say learning rate = 0.1):

$$
V_{\text{new}} = V - 0.1 \cdot \delta_{\text{output}} \cdot h
$$

This nudges the weights slightly so that next time the prediction moves closer to the truth.

---

## 🌟 Intuition

- **Forward pass:** “What’s my prediction with current weights?”
- **Backward pass:** “How wrong was I? Adjust knobs in the direction that reduces error.”

Repeat this many times over all data → the network “learns.”
