```python
x_train,x_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2,random_state=42)
```

```python
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
encoder=LabelEncoder()
y_train=encoder.fit_transform(y_train)
y_test=encoder.transform(y_test)
```

#NUMPY ARRAYS TO TENSORS FOR PYTORCH


```python
x_train_tensor=torch.from_numpy(x_train)
x_test_tensor=torch.from_numpy(x_test)
y_train_tensor=torch.from_numpy(y_train)
y_test_tensor=torch.from_numpy(y_test)
```

```python
class MySimpleNN():
  def __init__(self,x):
    self.weights=torch.rand(x.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias=torch.zeros(1,dtype=torch.float64,requires_grad=True)
  def forward(self,x):
    z=torch.matmul(x,self.weights)+self.bias
    y_pred=torch.sigmoid(z)
    return y_pred
  def loss_function(self,y_pred,y):
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred))
    loss = loss.mean()
    return loss
```

##TRAINING PIPELINE

```python
#create model
model=MySimpleNN(x_train_tensor)
model.weights
```

```python
model.bias
```

```python
learning_rate=0.01
epochs=1000
model=MySimpleNN(x_train_tensor)
#define loop
for epoch in range(epochs):
  #forward pass
  y_pred=model.forward(x_train_tensor)

  # loss calculate
  loss=model.loss_function(y_pred,y_train_tensor)

  #backward pass
  loss.backward()

  #parameters update
  with torch.no_grad():
    model.weights-=learning_rate*model.weights.grad.data
    model.bias-=learning_rate*model.bias.grad.data

  #zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  #print loss in each epoch
  print(f"epoch {epoch+1} loss {loss.item()}")
```

```python
# model evaluate
with torch.no_grad():
  y_pred=model.forward(x_test_tensor)

  y_pred=(y_pred>1).float()
  accuracy=(y_pred==y_test_tensor).float().mean()
  print(f"accuracy {accuracy}")

```

```python
class MySimpleNN(nn.Module):
  def __init__(self,input_size):
    super().__init__()
    self.linear = nn.Linear(input_size,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,x):
    out = self.linear(x)
    out = self.sigmoid(out)
    return out

```

```python
# define loss function
loss_function= nn.BCELoss()
```

```python
y_train_tensor.reshape(-1,1).float()
```

```python
learning_rate=0.01
epochs=1000
model=MySimpleNN(x_train_tensor.shape[1])
# define optimizer
optimizer = torch.optim.SGD(model.parameters(),lr=learning_rate)
#define loop
for epoch in range(epochs):
  #forward pass
  y_pred=model(x_train_tensor.float())# no need to call model.forward(), just call model(), forward function will call automatically due to magic function
  # loss calculate
  loss=loss_function(y_pred,y_train_tensor.view(-1,1).float())#.view() fucntion is similar to .reshape()

  # clear gradients
  optimizer.zero_grad()

  #backward pass
  loss.backward()

  #parameters update
  optimizer.step()

  #zero gradients
  model.linear.weight.grad.zero_()
  model.linear.bias.grad.zero_()

  #print loss in each epoch
  print(f"epoch {epoch+1} loss {loss.item()}")
```

```python
# model evaluate
with torch.no_grad():
  y_pred=model.forward(x_test_tensor.float())

  y_pred=(y_pred>1).float()
  accuracy=(y_pred==y_test_tensor).float().mean()
  print(f"accuracy {accuracy}")

```
