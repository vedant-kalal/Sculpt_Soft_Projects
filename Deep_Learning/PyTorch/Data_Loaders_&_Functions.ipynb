#PyTorch Data Loading
##1. Dataset Class

PyTorch represents data using the torch.utils.data.Dataset class.

It is an abstract class that you can customize.

Two key methods you must define:

__len__: Returns the size of the dataset.#
__getitem__: Returns a single sample (data + label) by index.

```python
from torch.utils.data import Dataset

class MyDataset(Dataset):
    def __init__(self, data, labels):
        self.data = data
        self.labels = labels

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx]
Dataset
```

#2. DataLoader Class

The torch.utils.data.DataLoader helps in batching, shuffling, and parallel loading of data.

##Main arguments:

dataset → the dataset object.


batch_size → number of samples per batch.


shuffle → whether to shuffle each epoch.


num_workers → number of subprocesses for loading data.

```python
from torch.utils.data import DataLoader

# Suppose dataset has 100 samples
dataset = MyDataset(data=range(100), labels=range(100))

loader = DataLoader(dataset, batch_size=10, shuffle=True)

for batch_data, batch_labels in loader:
    print(batch_data, batch_labels)
    break
```

#Summary

DataLoader → Efficiently batches, shuffles, and loads samples.

#Loading our own personal dataset


```python
import torch
from torch.utils.data import Dataset, DataLoader
import pandas as pd

class HouseDataset(Dataset):
    def __init__(self, csv_file):
        self.data = pd.read_csv(csv_file)

        # Features (area, bedrooms)
        self.X = self.data.iloc[:, :-1].values  # all cols except last
        # Target (price)
        self.y = self.data.iloc[:, -1].values   # last col

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        X = torch.tensor(self.X[idx], dtype=torch.float32)
        y = torch.tensor(self.y[idx], dtype=torch.float32)
        return X, y
```

```python
# Load dataset
dataset = HouseDataset("mnist_test.csv")

# Wrap in DataLoader (batching, shuffling)
loader = DataLoader(dataset, batch_size=2, shuffle=True)

# Iterate
for features, labels in loader:
    print("Features:", features)
    print("Labels:", labels)
    break
```
