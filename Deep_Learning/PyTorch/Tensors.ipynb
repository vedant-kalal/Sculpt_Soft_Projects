# 📘 Tensors in Deep Learning (with PyTorch)

## 1. What is a Tensor?
A **tensor** is a specialized multi-dimensional array designed for mathematical and computational efficiency. Think of it as a generalization of scalars, vectors, and matrices to higher dimensions. Tensors are the basic data structure in PyTorch (and other deep learning frameworks).

---

## 2. Types of Tensors with Real-World Examples

| Tensor Type | Dimensions | Shape Example         | Real-World Example                       |
|-------------|------------|----------------------|------------------------------------------|
| Scalar      | 0D         | ()                   | Loss value (e.g., 5.0, -3.14)            |
| Vector      | 1D         | (n,)                 | Word embedding [0.12, -0.84, 0.33]       |
| Matrix      | 2D         | (m, n)               | Grayscale image [[0,255,128],[34,90,180]] |
| 3D Tensor   | 3D         | (w, h, c)            | RGB image (e.g., 256×256×3)              |
| 4D Tensor   | 4D         | (batch, w, h, c)     | Batch of 32 RGB images: [32,128,128,3]   |
| 5D Tensor   | 5D         | (batch, t, w, h, c)  | Video clips: [10,16,64,64,3]             |

### Examples
- **Scalars (0D):** Single value, e.g., loss = 5.0
- **Vectors (1D):** Feature vector, e.g., [0.12, -0.84, 0.33]
- **Matrices (2D):** Grayscale image, e.g.,
  ```
  [[0, 255, 128],
   [34,  90, 180]]
  ```
- **3D Tensors:** RGB image, shape [256, 256, 3]
- **4D Tensors:** Batch of 32 RGB images, shape [32, 128, 128, 3]
- **5D Tensors:** Batch of 10 video clips, each with 16 frames of size 64×64 (RGB): [10, 16, 64, 64, 3]

---

## 3. Why Are Tensors Useful?

- **Mathematical Operations:**
  - Enable fast operations: addition, multiplication, dot product, etc.
  - Essential for neural network computations.
- **Representation of Real-world Data:**
  - Images: 3D tensor (width × height × channels)
  - Text: Tokenized into 2D or 3D tensor (sequence length × embedding size)
  - Audio: 2D tensor (time × frequency)
  - Videos: 5D tensor (batch × frames × width × height × channels)
- **Efficient Computations:**
  - Optimized for GPUs/TPUs
  - Enable large-scale matrix multiplications crucial for training deep models

---

## 4. Where Are Tensors Used in Deep Learning?

- **Data Storage:**
  - Training data (images, text, audio, videos) stored as tensors
- **Weights & Biases:**
  - Neural network parameters (learnable values) are stored as tensors
  - Example: W (weights matrix), b (bias vector)
- **Matrix Operations in Neural Networks:**
  - Forward pass = matrix multiplications using tensors
  - Example:  
    $z = W \cdot x + b$  
    where W, x, and b are all tensors
- **Training Process:**
  - Forward pass: Tensors flow through the network to produce predictions
  - Backward pass: Gradients (also tensors) are computed using automatic differentiation

---

## 5. Summary (Quick Recall)

- **Scalars:** 0D (single value)
- **Vectors:** 1D (list of numbers)
- **Matrices:** 2D (tables)
- **3D+ Tensors:** Images, videos, batches
- **Tensors power deep learning:** Store data, model parameters, and enable fast GPU computations

## Creating a Tensor

````python
# using empty
# Creates a tensor of shape (2, 3) with uninitialized values (whatever is in memory at that location).
# Useful for performance if you plan to fill the tensor with values right after creation.
a = torch.empty(2,3)
````

````python
# check type
# Returns the Python type of the variable 'a'.
# For a PyTorch tensor, this will be <class 'torch.Tensor'>.
type(a)
````

````python
# using zeros
# Creates a tensor of shape (2, 3) filled with zeros.
torch.zeros(2,3)
````

````python
# using ones
# Creates a tensor of shape (2, 3) filled with ones.
torch.ones(2,3)
````

````python
# using rand
# Creates a tensor of shape (2, 3) with random values uniformly sampled from [0, 1).
torch.rand(2,3)
````

````python
# use of seed
# Generates a tensor of shape (2, 3) with random values.
# Without setting a seed, the values will be different each time you run this cell.
torch.rand(2,3)
````

````python
# manual_seed
# Sets the seed for generating random numbers, so results are reproducible.
torch.manual_seed(100)
# Now, every time you run this, the random values will be the same.
torch.rand(2,3)
````

````python
# manual_seed
# Setting the seed again ensures reproducibility for the next random operation.
torch.manual_seed(100)
# Generates the same random tensor as before due to the same seed.
torch.rand(2,3)
````

````python
# using tensor
# Creates a tensor directly from a nested Python list (2x3 matrix).
torch.tensor([[1,2,3],[4,5,6]])
````

````python
# other ways

# arange
# Creates a 1D tensor with values from 0 to 8 (exclusive of 10), stepping by 2.
print("using arange ->", torch.arange(0,10,2))

# using linspace
# Creates a 1D tensor with 10 values evenly spaced between 0 and 10 (inclusive).
print("using linspace ->", torch.linspace(0,10,10))

# using eye
# Creates a 5x5 identity matrix (ones on the diagonal, zeros elsewhere).
print("using eye ->", torch.eye(5))

# using full
# Creates a 3x3 tensor filled with the value 5.
print("using full ->", torch.full((3, 3), 5))
````

## Mathematical operations

### 1. Scalar operation

````python
# Create a 2x2 tensor with random values between 0 and 1
x = torch.rand(2,2)
# Display the tensor
x
````

````python
# addition
# Adds 2 to every element in x
x + 2

# subtraction
# Subtracts 2 from every element in x
x - 2

# multiplication
# Multiplies every element in x by 3
x * 3

# division
# Divides every element in x by 3
x / 3

# integer division
# Multiplies x by 100, then performs integer division by 3 (drops decimals)
(x * 100)//3

# modulo
# Finds the remainder when each element (after integer division) is divided by 2
((x * 100)//3)%2

# power
# Squares every element in x (raises to the power of 2)
x**2
````

### 2. Element wise operation

````python
# Create two 2x3 tensors with random values between 0 and 1
a = torch.rand(2,3)
b = torch.rand(2,3)

# Print the tensors to see their values
print(a)
print(b)
````

````python
# add
# Adds corresponding elements of a and b
a + b

# sub
# Subtracts corresponding elements of b from a
a - b

# multiply
# Multiplies corresponding elements of a and b
a * b

# division
# Divides corresponding elements of a by b
a / b

# power
# Raises each element of a to the power of the corresponding element in b
a ** b

# mod
# Computes the remainder of each element in a divided by the corresponding element in b
a % b
````

````python
# Create a 1D tensor with both positive and negative values
c = torch.tensor([1, -2, 3, -4])
````

````python
# abs
# Returns a tensor with the absolute value of each element in c (removes negative sign)
torch.abs(c)
````

````python
# negative
# Returns a tensor with the sign of each element in c flipped (positive becomes negative and vice versa)
torch.neg(c)
````

````python
# Create a 1D tensor with floating-point values
d = torch.tensor([1.9, 2.3, 3.7, 4.4])
````

````python
# round
# Rounds each element in d to the nearest integer
torch.round(d)
````

````python
# ceil
# Rounds each element in d up to the nearest integer (ceiling)
torch.ceil(d)
````

````python
# floor
# Rounds each element in d down to the nearest integer (floor)
torch.floor(d)
````

````python
# clamp
# Restricts each element in d to be within the range [2, 3]
torch.clamp(d, min=2, max=3)
````

### 3. Reduction operation

````python
# Create a 2x3 tensor with random float values between 0 and 10
# Used for demonstrating reduction operations (sum, mean, etc.)
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float32)
e
````

````python
# sum
# Computes the sum of all elements in e
torch.sum(e)

# sum along columns (dim=0)
# Sums elements in each column, returns a 1D tensor with the sum for each column
torch.sum(e, dim=0)

# sum along rows (dim=1)
# Sums elements in each row, returns a 1D tensor with the sum for each row
torch.sum(e, dim=1)
````

````python
# mean
# Computes the mean (average) of all elements in e
torch.mean(e)

# mean along columns (dim=0)
# Computes the mean for each column
torch.mean(e, dim=0)
````

````python
# median
# Returns the median value of all elements in e (middle value when sorted)
torch.median(e)
````

````python
# max and min
# Returns the maximum value in e
torch.max(e)
# Returns the minimum value in e
torch.min(e)
````

````python
# product
# Returns the product of all elements in e (multiplies them together)
torch.prod(e)
````

````python
# standard deviation
# Returns the standard deviation of all elements in e (measure of spread)
torch.std(e)
````

````python
# variance
# Returns the variance of all elements in e (measure of how far values spread from the mean)
torch.var(e)
````

````python
# argmax
# Returns the index of the maximum value in e (flattened tensor)
torch.argmax(e)
````

````python
# argmin
# Returns the index of the minimum value in e (flattened tensor)
torch.argmin(e)
````

### 4. Matrix operations

````python
# Create a 2x3 matrix (tensor) with random integers between 0 and 9
f = torch.randint(size=(2,3), low=0, high=10)
# Create a 3x2 matrix (tensor) with random integers between 0 and 9
g = torch.randint(size=(3,2), low=0, high=10)

# Print both matrices to see their values
print(f)
print(g)
````

````python
# matrix multiplication
# Multiplies matrix f (2x3) with matrix g (3x2) using matrix multiplication rules
# The result is a 2x2 matrix
torch.matmul(f, g)
````

````python
# Create two 1D tensors (vectors)
vector1 = torch.tensor([1, 2])
vector2 = torch.tensor([3, 4])

# dot product
# Computes the dot product of vector1 and vector2 (1*3 + 2*4 = 11)
torch.dot(vector1, vector2)
````

````python
# transpose
# Swaps rows and columns of matrix f (2x3 becomes 3x2)
torch.transpose(f, 0, 1)
````

````python
# Create a 3x3 matrix (tensor) with random float values between 0 and 9
h = torch.randint(size=(3,3), low=0, high=10, dtype=torch.float32)
h
````

````python
# determinant
# Computes the determinant of matrix h (must be square)
torch.det(h)
````

````python
# inverse
# Computes the inverse of matrix h (must be square and non-singular)
torch.inverse(h)
````

### 5. Comparison operations

````python
# Create two 2x3 tensors with random integers between 0 and 9
i = torch.randint(size=(2,3), low=0, high=10)
j = torch.randint(size=(2,3), low=0, high=10)

# Print both tensors to see their values
print(i)
print(j)
````

````python
# greater than
# Returns a boolean tensor where each element is True if i > j, else False
i > j

# less than
# Returns a boolean tensor where each element is True if i < j, else False
i < j

# equal to
# Returns a boolean tensor where each element is True if i == j, else False
i == j

# not equal to
# Returns a boolean tensor where each element is True if i != j, else False
i != j

# greater than or equal to
# Returns a boolean tensor where each element is True if i >= j, else False
i >= j

# less than or equal to
# Returns a boolean tensor where each element is True if i <= j, else False
i <= j
````

## Inplace Operations

````python
# Create two 2x3 tensors with random values between 0 and 1
m = torch.rand(2,3)
n = torch.rand(2,3)

# Print both tensors to see their values
print(m)
print(n)
````

````python
# In-place addition: adds n to m and stores the result in m
# The original m is modified (not a new tensor)
m.add_(n)
````

````python
# Display the modified tensor m after in-place addition
m
````

````python
# Display the tensor n (unchanged)
n
````

````python
# Applies the ReLU function to m (not in-place)
# Returns a new tensor where all negative values in m are set to 0
torch.relu(m)
````

````python
# In-place ReLU: sets all negative values in m to 0, modifying m directly
m.relu_()
````

````python
# Display the modified tensor m after in-place ReLU
m
````

## Copying a Tensor

````python
# Create a 2x3 tensor with random values between 0 and 1
a = torch.rand(2,3)
# Display the tensor a
a
````

````python
# Assign a to b (no copy, just reference)
b = a
````

````python
# Display b (which is just a reference to a)
b
````

````python
# Modify the first element of a (also changes b since b is just a reference)
a[0][0] = 0
````

````python
# Display a after modification
a
````

````python
# Display b after modification of a (shows b is also changed)
b
````

````python
# Check the memory id of a (shows both a and b point to the same object)
id(a)
````

````python
# Check the memory id of b (should be the same as a)
id(b)
````

````python
# Make a true copy of a using clone()
b = a.clone()
````

````python
# Display a after cloning (should be unchanged)
a
````

````python
# Display b after cloning (should be a separate copy)
b
````

````python
# Modify a after cloning (should not affect b)
a[0][0] = 10
````

````python
# Display a after modification (should show the change)
a
````

````python
# Display b after a is modified (should be unchanged if clone worked)
b
````

````python
# Check the memory id of a (should be different from b after clone)
id(a)
````

````python
# Check the memory id of b (should be different from a after clone)
id(b)
````
