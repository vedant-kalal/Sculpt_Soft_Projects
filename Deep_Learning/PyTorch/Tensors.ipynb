# 📘 Tensors in Deep Learning (with PyTorch)

## 1. What is a Tensor?
A **tensor** is a specialized multi-dimensional array designed for mathematical and computational efficiency. Think of it as a generalization of scalars, vectors, and matrices to higher dimensions. Tensors are the basic data structure in PyTorch (and other deep learning frameworks).

---

## 2. Types of Tensors with Real-World Examples

| Tensor Type | Dimensions | Shape Example         | Real-World Example                       |
|-------------|------------|----------------------|------------------------------------------|
| Scalar      | 0D         | ()                   | Loss value (e.g., 5.0, -3.14)            |
| Vector      | 1D         | (n,)                 | Word embedding [0.12, -0.84, 0.33]       |
| Matrix      | 2D         | (m, n)               | Grayscale image [[0,255,128],[34,90,180]] |
| 3D Tensor   | 3D         | (w, h, c)            | RGB image (e.g., 256×256×3)              |
| 4D Tensor   | 4D         | (batch, w, h, c)     | Batch of 32 RGB images: [32,128,128,3]   |
| 5D Tensor   | 5D         | (batch, t, w, h, c)  | Video clips: [10,16,64,64,3]             |

### Examples
- **Scalars (0D):** Single value, e.g., loss = 5.0
- **Vectors (1D):** Feature vector, e.g., [0.12, -0.84, 0.33]
- **Matrices (2D):** Grayscale image, e.g.,
  ```
  [[0, 255, 128],
   [34,  90, 180]]
  ```
- **3D Tensors:** RGB image, shape [256, 256, 3]
- **4D Tensors:** Batch of 32 RGB images, shape [32, 128, 128, 3]
- **5D Tensors:** Batch of 10 video clips, each with 16 frames of size 64×64 (RGB): [10, 16, 64, 64, 3]

---

## 3. Why Are Tensors Useful?

- **Mathematical Operations:**
  - Enable fast operations: addition, multiplication, dot product, etc.
  - Essential for neural network computations.
- **Representation of Real-world Data:**
  - Images: 3D tensor (width × height × channels)
  - Text: Tokenized into 2D or 3D tensor (sequence length × embedding size)
  - Audio: 2D tensor (time × frequency)
  - Videos: 5D tensor (batch × frames × width × height × channels)
- **Efficient Computations:**
  - Optimized for GPUs/TPUs
  - Enable large-scale matrix multiplications crucial for training deep models

---

## 4. Where Are Tensors Used in Deep Learning?

- **Data Storage:**
  - Training data (images, text, audio, videos) stored as tensors
- **Weights & Biases:**
  - Neural network parameters (learnable values) are stored as tensors
  - Example: W (weights matrix), b (bias vector)
- **Matrix Operations in Neural Networks:**
  - Forward pass = matrix multiplications using tensors
  - Example:  
    $z = W \cdot x + b$  
    where W, x, and b are all tensors
- **Training Process:**
  - Forward pass: Tensors flow through the network to produce predictions
  - Backward pass: Gradients (also tensors) are computed using automatic differentiation

---

## 5. Summary (Quick Recall)

- **Scalars:** 0D (single value)
- **Vectors:** 1D (list of numbers)
- **Matrices:** 2D (tables)
- **3D+ Tensors:** Images, videos, batches
- **Tensors power deep learning:** Store data, model parameters, and enable fast GPU computations
