# 🔹 Overview

**PyTorch** is an open-source deep learning library developed by Meta AI (formerly Facebook AI Research).

- **Combines:**
  - **Python:** Ease of use
  - **Torch (Lua-based):** High-performance tensor computations, GPU acceleration
- **Known for:**
  - Dynamic computation graphs (eager execution)
  - Intuitive Pythonic API
  - Strong adoption in research and academia

---

# 🔹 Key Features of PyTorch

- **Tensor Computations:** Similar to NumPy but with GPU support
- **GPU Acceleration:** Seamless CUDA integration
- **Dynamic Computation Graph:** Eager execution (define-by-run)
- **Automatic Differentiation:** `torch.autograd` for backprop
- **Distributed Training:** `torch.distributed` across GPUs/nodes
- **Interoperability:** Works well with NumPy, SciPy, scikit-learn

# 🔹 PyTorch vs TensorFlow

| Aspect              | PyTorch                                      | TensorFlow                                 | Verdict                                      |
|---------------------|----------------------------------------------|--------------------------------------------|----------------------------------------------|
| Programming Language| Python-first                                 | Python, C++, Java, JS, Swift               | Depends: PyTorch for Python, TF for multi-language |
| Ease of Use         | Intuitive, Pythonic                          | TF 2.x improved, still complex             | PyTorch wins                                 |
| Deployment          | TorchScript, PyTorch Mobile                  | TF Serving, Lite, JS                       | TensorFlow wins                              |
| Performance         | JIT + TorchScript, competitive               | XLA compiler, static graph                 | Tie                                          |
| Community           | Research-focused                             | Large industry adoption                    | Depends                                      |
| High-Level APIs     | torch.nn, Lightning, Fastai                  | Keras (tf.keras)                           | TensorFlow wins                              |
| Mobile Deployment   | PyTorch Mobile                               | TF Lite, TF.js                             | TensorFlow wins                              |
| Preferred Domains   | Academia, rapid prototyping                  | Industry, production                       | Depends                                      |
| Learning Curve      | Easier, dynamic                              | Steeper, static (but better in TF 2.x)     | PyTorch wins                                 |
| Interoperability    | Great with Python ecosystem, ONNX            | TF Hub, SavedModel                         | PyTorch wins                                 |
| Customizability     | High, flexible                               | Less flexible, improved in TF 2.x          | PyTorch wins                                 |
| Deployment Tools    | TorchServe, cloud support                    | TF Serving, TFX pipelines                  | TensorFlow wins                              |
| Distributed Training| torch.distributed, Horovod                   | tf.distribute.Strategy                     | TensorFlow wins                              |
| Model Zoo           | TorchVision, Hugging Face                    | TF Hub                                     | Tie                                          |

# 🔹 PyTorch Core Modules

| Module                | Description                                               |
|-----------------------|----------------------------------------------------------|
| `torch`               | Core tensor operations (multidimensional arrays + math)  |
| `torch.autograd`      | Automatic differentiation (computes gradients)           |
| `torch.nn`            | Neural network layers, activations, loss functions       |
| `torch.optim`         | Optimization algorithms (SGD, Adam, RMSProp)            |
| `torch.utils.data`    | Dataset & DataLoader for efficient data handling         |
| `torch.jit`           | JIT compilation, TorchScript for deployment              |
| `torch.distributed`   | Distributed training on multiple GPUs/machines           |
| `torch.cuda`, `torch.backends` | GPU acceleration & backend control (cuDNN, MKL) |
| `torch.multiprocessing`| Parallelism with CUDA tensors                           |
| `torch.quantization`  | Model quantization (smaller, faster models)              |
| `torch.onnx`          | Export models to ONNX format                             |

# 🔹 PyTorch Journey Summary

- Started in research labs (dynamic graphs, Pythonic design)
- Evolved to support production deployment (TorchScript, Mobile, ONNX)
- Became industry-ready with PyTorch 2.0 → speed, scalability, hardware optimization
- Now dominates research community, while TensorFlow remains strong in enterprise production
