```python
df_amenities = df['Amenities'].str.get_dummies(sep=',')

# Clean column names
df_amenities.columns = df_amenities.columns.str.strip().str.lower().str.replace(' ', '_')

# Merge back
df = pd.concat([df, df_amenities], axis=1)
```

```python
df.drop(columns=['ID', 'Property Type','Property Name'], inplace=True)
df.head()
```

```python
property_effect = df.groupby('Address Area')['Super Buildup Rate (per sq.ft.)'].mean().sort_values()

print(property_effect.head())
print(property_effect.tail())
```

```python
df.head()
```

```python
from category_encoders import TargetEncoder

te = TargetEncoder(cols=['Address Area'])
df['Address_Area_encoded'] = te.fit_transform(df['Address Area'], df['Super Buildup Rate (per sq.ft.)'])
```

```python
config_map = {
    "1BHK": 1, "2BHK": 2, "3BHK": 3,
    "4BHK": 4, "Duplex": 5, "Penthouse": 6
}
df['Configuration_encoded'] = df['Configuration'].map(config_map)
```

```python
condition_map = {
    "Requires Renovation": 1,
    "Good": 2,
    "Fair": 3,
    "New": 4
}
df['Condition_encoded'] = df['Condition'].map(condition_map)
```

```python
df = pd.get_dummies(df, columns=['Landmark'], prefix='Landmark')
```

```python
# Calculate ratio (last / first)
df["Trend_Ratio"] = df["Rate 2025 (per sq.ft.)"] / df["Rate 2021 (per sq.ft.)"]

# Normalize ratio into [0,1]
# Using logistic/sigmoid scaling so very high ratios don’t explode
df["Trend_Score"] = 1 / (1 + np.exp(-(df["Trend_Ratio"] - 1) * 5))
```

```python
from sklearn.preprocessing import StandardScaler

# Columns to normalize (large values)
cols_to_normalize = [
    'Carpet Area (sq.ft.)',
    'Super Buildup Rate (per sq.ft.)',
    'Built-Up Area (sq.ft.)',
    'Area as per Approved Plan (sq.ft.)',
    'Valuation Area (sq.ft.)',
    'Super Built-up Area (sq.ft.)',
    'Total Value (INR)',
    'Value of Car Park (INR)',
    'Final Value by Comparison Method (INR)',
    'Last Sale Price (per sq.ft.)',
    'Last to Last Sale Price (per sq.ft.)'
]

# Initialize scaler
scaler = StandardScaler()

# Fit-transform only on the selected columns
df[cols_to_normalize] = scaler.fit_transform(df[cols_to_normalize])
```

```python
# List of columns to drop (already encoded or duplicated)
drop_cols = [
    "Amenities",
    "Address Area",
    "Configuration",
    "Condition",
    "Landmark Distance (KM)",
    "Rate 2021 (per sq.ft.)", "Rate 2022 (per sq.ft.)", 
    "Rate 2023 (per sq.ft.)", "Rate 2024 (per sq.ft.)", "Rate 2025 (per sq.ft.)"
]


# Drop unnecessary columns
df_cleaned = df.drop(columns=drop_cols, errors="ignore")

# Save to new CSV
df_cleaned.to_csv("cleaned_property_data.csv", index=False)
```
