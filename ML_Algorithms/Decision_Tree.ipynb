# Decision Tree: Intuitive Notes

## 1. What is a Decision Tree?
- Think of it like the 20 Questions game.
- At the top (root), you ask the most important question.
- Depending on the answer, you move down the tree (yes/no or category).
- At the bottom (leaf), you make a decision (class label or value).

**Example (loan approval):**
- Root Q: “Is income > 50k?”
  - If yes → next Q: “Has good credit history?”
    - If yes → approve loan
    - If no → reject loan
  - If no → reject loan
- That’s literally a Decision Tree.

## 2. Types of Decision Trees
- **Classification Tree:** predicts categories (e.g., spam / not spam).
- **Regression Tree:** predicts continuous values (e.g., house price).

## 3. The Math Behind It (How are questions chosen?)
We need a way to measure which feature is the best split.

### 1. Entropy (measure of impurity/randomness)
$$
Entropy(S) = -\sum_{i=1}^c p_i \log_2(p_i)
$$
- $p_i$ = probability of class $i$ in set $S$.
- If all samples are the same class → Entropy = 0 (pure).
- If samples are 50-50 → Entropy = 1 (max uncertainty).

### 2. Information Gain (IG)
$$
IG(S, A) = Entropy(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} Entropy(S_v)
$$
- $S$ = dataset, $A$ = attribute (feature), $S_v$ = subset after splitting on value $v$.
- Bigger IG = better split.

### 3. Gini Index (alternative to entropy)
$$
Gini(S) = 1 - \sum_{i=1}^c p_i^2
$$
- Used in CART (Classification and Regression Trees).
- Lower Gini = better purity.

## 4. How Tree is Built (Algorithm: ID3 / CART)
1. Start with all data.
2. Calculate Entropy (or Gini).
3. For each feature, check Information Gain.
4. Pick the feature with highest IG → root.
5. Split dataset based on that feature.
6. Repeat until:
   - All leaves are pure (one class), OR
   - Max depth reached, OR
   - No improvement possible.

## 5. Example (Simple Binary Dataset)
**Dataset: Predict Play Tennis based on Weather.**
| Weather   | Play |
|-----------|------|
| Sunny     | No   |
| Sunny     | No   |
| Overcast  | Yes  |
| Rain      | Yes  |
| Rain      | No   |
| Overcast  | Yes  |

- Step 1: Check Entropy before split.
- Step 2: Try splitting on "Weather" → calculate IG.
- Step 3: “Overcast” is pure (all Yes), “Sunny” is mostly No, “Rain” is mixed. So root = Weather.
- **Tree:**
  - If Overcast → Play=Yes
  - If Sunny → Play=No
  - If Rain → Ask another question…

## 6. Pros & Cons
**Pros:**
- Easy to understand (like flowcharts).
- Works on categorical + numerical data.
- No feature scaling needed.

**Cons:**
- Can overfit (too deep trees).
- Small changes in data → different tree.
- Greedy algorithm (locally best split, not always globally best).
- That’s why in practice we use Random Forests or Gradient Boosted Trees (they fix overfitting).

## 7. Quick Visual Mnemonic
- **Entropy:** "How mixed is it?"
- **Info Gain:** "How much cleaner after split?"
- **Tree growth:** "Ask the most useful question first."
