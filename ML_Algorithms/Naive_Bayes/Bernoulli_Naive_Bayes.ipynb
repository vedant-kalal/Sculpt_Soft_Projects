# Bernoulli Naive Bayes: Intuitive Notes

## 1. What is Bernoulli Naive Bayes?
- A special version of Naive Bayes for **binary features** (0/1, yes/no, present/absent).
- Example: Does an email contain the word “win”? → Yes (1) or No (0).
- Works well for text classification when only presence/absence of words matters (not their count).

## 2. The Formula
- For a class $C_i$ and binary features $x = (x_1, x_2, ..., x_n)$:
  $P(C_i|x) \propto P(C_i) \cdot \prod_{j=1}^n P(x_j|C_i)^{x_j} \cdot (1 - P(x_j|C_i))^{1-x_j}$
- **Interpretation:**
  - If $x_j = 1$ (feature present), use $P(x_j|C_i)$.
  - If $x_j = 0$ (feature absent), use $1 - P(x_j|C_i)$.
- The model accounts for both presence and absence of each feature.

## 3. Example: Spam Detection
| Contains “Win” | Contains “Free” | Class     |
|---------------|----------------|-----------|
| 1             | 1              | Spam      |
| 1             | 0              | Spam      |
| 0             | 1              | Spam      |
| 0             | 0              | Not Spam  |

- **Step 1: Priors**  
  $P(Spam) = \frac{3}{4}$, $P(NotSpam) = \frac{1}{4}$
- **Step 2: Likelihoods**
  - From Spam emails (3 rows):
    - $P(Win=1|Spam) = \frac{2}{3}$
    - $P(Free=1|Spam) = \frac{2}{3}$
  - From NotSpam (1 row):
    - $P(Win=1|NotSpam) = 0$
    - $P(Free=1|NotSpam) = 0$
- **Step 3: New Email → (Win=1, Free=0)**
  - $P(Spam|x) \propto P(Spam) \cdot P(Win=1|Spam) \cdot (1 - P(Free=1|Spam))$
  - $= \frac{3}{4} \cdot \frac{2}{3} \cdot (1 - \frac{2}{3}) = 0.25$
  - $P(NotSpam|x) \propto \frac{1}{4} \cdot 0 \cdot (1-0) = 0$
  - **Prediction = Spam**

## 4. Why Use Bernoulli NB?
- Good for binary data (yes/no, present/absent).
- Works well in text classification when only presence/absence matters.
- Simpler than Multinomial NB.

## 5. Limitations
- Ignores frequency of words (e.g., if “win” appears 10 times, it still counts as 1).
- If probabilities are 0, predictions can collapse → need Laplace smoothing.

## 6. Full Bernoulli Naive Bayes Formula
- **Bayes’ Rule:**  
  $P(C_i|x) = \frac{P(C_i) \cdot P(x|C_i)}{P(x)}$
- Since $P(x)$ is the same for all classes:  
  $P(C_i|x) \propto P(C_i) \cdot P(x|C_i)$
- **Bernoulli Likelihood:**  
  $P(x|C_i) = \prod_{j=1}^n P(x_j=1|C_i)^{x_j} \cdot (1 - P(x_j=1|C_i))^{1-x_j}$
- **Full formula:**  
  $P(C_i|x) \propto P(C_i) \cdot \prod_{j=1}^n \left[ P(x_j=1|C_i)^{x_j} \cdot (1 - P(x_j=1|C_i))^{1-x_j} \right]$

- **Explanation of Terms:**
  - $P(C_i)$: prior probability of class $C_i$
  - $x_j$: the $j$-th feature (0 or 1)
  - $P(x_j=1|C_i)$: probability that feature $j$ is present given class $C_i$
  - $1 - P(x_j=1|C_i)$: probability that feature $j$ is absent given class $C_i$

## ✅ Summary
- **Bernoulli Naive Bayes = Naive Bayes adapted for binary features, using both presence and absence probabilities.**
- Useful for binary data and text classification where only presence/absence matters.

## 📊 Comparison of Naive Bayes Variants
| Model           | Best For                                      | Likelihood Formula                                                                                                                                         | Advantages                                         | Limitations                                      |
|-----------------|-----------------------------------------------|-----------------------------------------------------------------------------------------------------------------------------------------------------------|----------------------------------------------------|--------------------------------------------------|
| **Bernoulli NB**| Binary features (present/absent).<br>Example: "Does email contain word ‘offer’?" (Yes/No) | $P(x \mid C_i) = \prod_j P(x_j=1\mid C_i)^{x_j} (1-P(x_j=1\mid C_i))^{1-x_j}$ | Simple for binary data; fast for text presence.    | Ignores frequency; needs Laplace smoothing.       |
| **Multinomial NB**| Discrete counts (word frequencies).<br>Example: text classification. | $P(x \mid C_i) = \frac{(\sum_j x_j)!}{\prod_j x_j!} \prod_j P(x_j\mid C_i)^{x_j}$ | Works well with document-term matrices; efficient. | Not for continuous or binary-only features.       |
| **Gaussian NB** | Continuous features (heights, scores, etc.)    | $P(x_j\mid C_i) = \frac{1}{\sqrt{2\pi\sigma_{ij}^2}} \exp\left(-\frac{(x_j-\mu_{ij})^2}{2\sigma_{ij}^2}\right)$ | Handles continuous data naturally; simple.         | Assumes Gaussian distribution; not for skewed data.|
| **Complement NB**| Imbalanced datasets (fraud, rare classes).     | Similar to Multinomial, but uses complement class counts:<br>$P(x\mid C_i) = \prod_j \left(\frac{\text{count}(word_j \in \text{NOT-}C_i)+\alpha}{\sum_k \text{count}(word_k \in \text{NOT-}C_i)+\alpha n}\right)^{x_j}$ | Handles imbalance better; robust for text.         | More complex; not needed for balanced data.       |
| **Categorical NB**| Categorical features (color=Red/Blue/Green).   | $P(x_j=k\mid C_i) = \frac{\text{count}(x_j=k, C_i)+\alpha}{\text{count}(C_i)+\alpha m_j}$ | Naturally handles categorical (non-numeric) data.  | Struggles with high-cardinality categories.       |

**Quick Memory Hook:**<br>
- Bernoulli → "Presence/Absence"<br>
- Multinomial → "Counts"<br>
- Gaussian → "Continuous"<br>
- Complement → "Imbalance Fix"<br>
- Categorical → "Categories"
