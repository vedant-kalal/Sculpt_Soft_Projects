# Naive Bayes: Intuitive Notes

## 1. What is Naive Bayes?
- A **probabilistic classifier** based on Bayes’ Theorem.
- Predicts the class of a new example by computing probabilities.
- Called “naive” because it assumes features are independent (rarely true, but works well in practice).

## 2. Bayes’ Theorem
- The core formula:  
  $P(A|B) = \frac{P(B|A) \cdot P(A)}{P(B)}$
- $P(A|B)$: Posterior = probability of class $A$ given evidence $B$.
- $P(A)$: Prior = initial probability of $A$.
- $P(B|A)$: Likelihood = probability of seeing $B$ if $A$ is true.
- $P(B)$: Evidence (normalizing constant).

## 3. For Classification
- Suppose we want to classify $x = (x_1, x_2, ..., x_n)$ into one of classes $C_1, C_2, ..., C_k$.
- We want:  
  $\hat{y} = \arg\max_{C_i} P(C_i|x)$
- Using Bayes:  
  $P(C_i|x) = \frac{P(x|C_i) \cdot P(C_i)}{P(x)}$
- Since $P(x)$ is the same for all classes, we maximize:  
  $P(C_i|x) \propto P(C_i) \cdot P(x|C_i)$

- **Naive assumption (features independent):**  
  $P(x|C_i) = \prod_{j=1}^n P(x_j|C_i)$
- So:  
  $P(C_i|x) \propto P(C_i) \cdot \prod_{j=1}^n P(x_j|C_i)$

## 4. Types of Naive Bayes
- **Bernoulli NB:** Features are binary (yes/no, 0/1).
- **Multinomial NB:** Features are counts (e.g., word counts in text).
- **Gaussian NB:** Features are continuous and assumed Gaussian (bell curve).

## 5. Example (Spam Detection)
| Email contains "Win"? | Email contains "Money"? | Spam/Not Spam |
|----------------------|------------------------|---------------|
| Yes                  | Yes                    | Spam          |
| Yes                  | No                     | Spam          |
| No                   | Yes                    | Spam          |
| No                   | No                     | Not Spam      |

- Step 1: Compute priors:  
  $P(Spam) = \frac{3}{4}$, $P(NotSpam) = \frac{1}{4}$
- Step 2: Compute likelihoods:  
  $P(Win=Yes|Spam) = \frac{2}{3}$, $P(Money=Yes|Spam) = \frac{2}{3}$  
  $P(Win=Yes|NotSpam) = 0$, $P(Money=Yes|NotSpam) = 0$  
  (Usually apply Laplace smoothing to avoid zeros.)
- Step 3: Predict new email: "Win + Money" (both yes):
  - $P(Spam|Win, Money) \propto P(Spam) \cdot P(Win|Spam) \cdot P(Money|Spam) = \frac{3}{4} \cdot \frac{2}{3} \cdot \frac{2}{3} = 0.33$
  - $P(NotSpam|Win, Money) \propto P(NotSpam) \cdot 0 \cdot 0 = 0$
- So prediction = Spam.

## 6. Why the Naive Assumption?
- Calculating $P(x|C_i)$ for all features together is hard (requires lots of data).
- **Naive Bayes assumes all features are conditionally independent given the class:**
  - $P(x_1, x_2, x_3|C_i) = P(x_1|C_i) \cdot P(x_2|C_i) \cdot P(x_3|C_i)$
- This simplifies the math and makes the model practical for high-dimensional data.

## 7. Plugging It Back
- Bayes’ formula becomes:  
  $P(C_i|x) \propto P(C_i) \cdot \prod_{j=1}^n P(x_j|C_i)$
- $P(C_i)$: prior probability of the class.
- $\prod_{j=1}^n P(x_j|C_i)$: product of probabilities of each feature given the class.
- This is why it’s called “Naive”: in reality, features are often correlated, but we pretend they are independent.

## 8. Simple Example (Step by Step)
Suppose we want to classify if someone likes cricket ($C$ = Yes/No) based on:
- $x_1$: wears jersey (yes/no)
- $x_2$: watches sports channel (yes/no)
Naive Bayes says:
  $P(Yes|x_1, x_2) \propto P(Yes) \cdot P(x_1|Yes) \cdot P(x_2|Yes)$
We multiply the individual probabilities instead of computing the joint probability.

## 9. Advantages
- Very fast, works well with large data.
- Great for text classification (spam, sentiment, etc.).

## 10. Limitations
- Assumes features are independent (not always true).
- Struggles if features are highly correlated.

## ✅ Summary
- Naive Bayes = Bayes’ theorem + independence assumption + counts/likelihoods.
- It chooses the class with the highest posterior probability.
- Instead of storing one giant probability for all features combined, Naive Bayes breaks it into smaller pieces, multiplies them, and predicts based on the largest posterior probability.
