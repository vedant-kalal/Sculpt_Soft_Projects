# Multinomial Naive Bayes: Intuitive Notes

## 1. When Do We Use Multinomial NB?
- Used when features are **counts** (e.g., how many times each word appears in a document).
- Example: In an email:
  - "win" appears 3 times,
  - "money" appears 2 times,
  - "free" appears 0 times.
- Unlike Bernoulli NB (which only cares about presence/absence), Multinomial NB considers frequency.

## 2. Formula for Likelihood
- For class $C_i$, feature vector $x = (x_1, x_2, ..., x_n)$ where $x_j$ = count of feature $j$:
  $P(x|C_i) = \frac{(\sum_j x_j)!}{\prod_j (x_j!)} \prod_{j=1}^n \theta_{ij}^{x_j}$
- $\theta_{ij} = P(x_j|C_i)$ = probability of feature $j$ in class $C_i$.
- The factorial part is a constant for classification, so we usually simplify:
  $P(x|C_i) \propto \prod_{j=1}^n \theta_{ij}^{x_j}$

## 3. Full Multinomial NB Formula
- $P(C_i|x) \propto P(C_i) \cdot \prod_{j=1}^n \theta_{ij}^{x_j}$
- Where:
  - $P(C_i)$ = prior probability of class.
  - $\theta_{ij} = \frac{\text{count of word } j \text{ in class } C_i + \alpha}{\sum_k (\text{count of word } k \text{ in class } C_i) + \alpha n}$
  - $\alpha$ = Laplace smoothing parameter (to avoid zero probabilities).

## 4. Example (Tiny Text Classification)
Suppose we want to classify messages as Spam or Not Spam, vocabulary = {“win”, “free”}.

**Training data:**
- Spam messages:
  - “win free win” → counts: {win=2, free=1}
  - “free free” → counts: {win=0, free=2}
- Not Spam:
  - “win game” → counts: {win=1, free=0}

**Step 1: Priors**
- $P(Spam) = \frac{2}{3}$, $P(NotSpam) = \frac{1}{3}$

**Step 2: Likelihoods (with Laplace smoothing $\alpha=1$)**
- For Spam:
  - total word count = (2+1+0+2) = 5
  - $P(win|Spam) = \frac{2+1}{5+2} = \frac{3}{7}$
  - $P(free|Spam) = \frac{3+1}{5+2} = \frac{4}{7}$
- For NotSpam:
  - total word count = 1
  - $P(win|NotSpam) = \frac{1+1}{1+2} = \frac{2}{3}$
  - $P(free|NotSpam) = \frac{0+1}{1+2} = \frac{1}{3}$

**Step 3: New Message → “win free”**
- Feature vector = {win=1, free=1}
- For Spam:
  - $P(Spam|x) \propto P(Spam) \cdot (\frac{3}{7})^1 \cdot (\frac{4}{7})^1 = \frac{2}{3} \cdot \frac{12}{49}$
- For NotSpam:
  - $P(NotSpam|x) \propto P(NotSpam) \cdot (\frac{2}{3})^1 \cdot (\frac{1}{3})^1 = \frac{1}{3} \cdot \frac{2}{9}$
- Spam score > NotSpam score → **Classify as Spam**.

## 5. Key Differences vs Bernoulli NB
- **Bernoulli NB:** Works on binary presence/absence.
- **Multinomial NB:** Works on word counts/frequencies.
- Both assume features are conditionally independent given the class.

## ✅ Proper Formula
$P(C_i|x) \propto P(C_i) \cdot \prod_{j=1}^n \theta_{ij}^{x_j}$

## 6. Advantages
- Very fast and efficient for high-dimensional data (like text).
- Works well for document classification, spam detection, and sentiment analysis.
- Handles word frequency information, not just presence/absence.
- Simple to implement and interpret.

## 7. Limitations
- Assumes features (words) are conditionally independent given the class (not always true in real text).
- Assumes word order does not matter (bag-of-words model).
- Can perform poorly if word counts are not informative for the task.
- Sensitive to zero probabilities (requires Laplace smoothing).
