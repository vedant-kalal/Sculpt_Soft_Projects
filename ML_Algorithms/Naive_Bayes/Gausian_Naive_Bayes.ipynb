# Gaussian Naive Bayes (GNB): Intuitive Notes

## 1. When Do We Use Gaussian NB?
- Use when features are **continuous** (not binary, not counts).
- Example: Predicting flower species based on petal length/width (continuous values).
- Instead of using counts, we assume each feature follows a **Normal (Gaussian) distribution**.

## 2. Bayes’ Rule Reminder
$P(C_i|x) \propto P(C_i) \cdot P(x|C_i)$
where $x = (x_1, x_2, ..., x_n)$.

## 3. Gaussian Likelihood Formula
Each feature $x_j$ given class $C_i$ is assumed to follow a Gaussian distribution:
$$
P(x_j|C_i) = \frac{1}{\sqrt{2\pi\sigma_{ij}^2}} \exp\left(-\frac{(x_j - \mu_{ij})^2}{2\sigma_{ij}^2}\right)
$$
- $\mu_{ij}$ = mean of feature $j$ in class $C_i$
- $\sigma_{ij}^2$ = variance of feature $j$ in class $C_i$

## 4. Full Gaussian Naive Bayes Formula
Since features are assumed independent:
$$
P(C_i|x) \propto P(C_i) \cdot \prod_{j=1}^n \frac{1}{\sqrt{2\pi\sigma_{ij}^2}} \exp\left(-\frac{(x_j - \mu_{ij})^2}{2\sigma_{ij}^2}\right)
$$

## 5. Step-by-Step Example
Suppose we classify whether a student Passes or Fails based on Hours Studied ($x$).

**Training data:**
- Class Pass: {6, 7, 8 hours}
- Class Fail: {1, 2, 3 hours}

**Step 1: Priors**
- $P(Pass) = \frac{3}{6} = 0.5$
- $P(Fail) = \frac{3}{6} = 0.5$

**Step 2: Compute Mean & Variance**
- For Pass: $\mu=7$, $\sigma^2=1$
- For Fail: $\mu=2$, $\sigma^2=1$

**Step 3: Predict New Student ($x = 4$ hours)**
- For Pass:
  $P(x=4|Pass) = \frac{1}{\sqrt{2\pi(1)}} \exp\left(-\frac{(4-7)^2}{2(1)}\right) = \frac{1}{\sqrt{2\pi}} e^{-4.5}$
- For Fail:
  $P(x=4|Fail) = \frac{1}{\sqrt{2\pi(1)}} \exp\left(-\frac{(4-2)^2}{2(1)}\right) = \frac{1}{\sqrt{2\pi}} e^{-2}$

**Step 4: Multiply by Prior**
- $P(Pass|x=4) \propto 0.5 \cdot \frac{1}{\sqrt{2\pi}} e^{-4.5}$
- $P(Fail|x=4) \propto 0.5 \cdot \frac{1}{\sqrt{2\pi}} e^{-2}$
- Fail probability is higher → Predict Fail.

## 6. Key Idea
- **Bernoulli NB:** binary features (present/absent)
- **Multinomial NB:** discrete counts (word frequencies)
- **Gaussian NB:** continuous values (heights, weights, scores)
