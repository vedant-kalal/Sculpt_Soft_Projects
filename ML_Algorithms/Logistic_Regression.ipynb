# Logistic Regression: Intuitive Notes

## 1. What is Logistic Regression?
- Despite the name, **logistic regression is used for classification** (e.g., yes/no, 0/1, spam/not spam, disease/healthy).
- **Input:** Features (e.g., height, weight, hours studied).
- **Output:** Probability (between 0 and 1).
- **Decision:** Classify into 0 or 1 (using a threshold, usually 0.5).

## 2. Why Not Linear Regression?
- Linear regression can predict values outside the [0, 1] range (e.g., -2 or 3.7), which doesn't make sense for probabilities.
- **Solution:** Use a special function called the **Sigmoid function** to squash outputs to (0, 1).

## 3. The Sigmoid Function (Core Idea)
- Formula:  
  $\sigma(z) = \frac{1}{1 + e^{-z}}$
- **Input:** $z$ can be any real number.
- **Output:** Always between 0 and 1.
- In logistic regression:  
  $P(y=1|x) = \sigma(w^T x + b)$
    - $w$ = weights (parameters to learn)
    - $b$ = bias
    - $x$ = input vector (features)

## 4. Decision Rule
- $\hat{y} = \begin{cases} 1 & \text{if } P(y=1|x) \geq 0.5 \\ 0 & \text{if } P(y=1|x) < 0.5 \end{cases}$

## 5. Cost Function (Mathematics)
- We train the model by finding the best $w, b$.
- **Loss function:** Log Loss (Binary Cross Entropy):
  $J(w, b) = -\frac{1}{m} \sum_{i=1}^m \left[ y^{(i)} \log(\hat{y}^{(i)}) + (1 - y^{(i)}) \log(1 - \hat{y}^{(i)}) \right]$
- If prediction matches true label → cost is small.
- If prediction is wrong → cost is large.
- This function is convex, so gradient descent works well.

## 6. Gradient Descent Updates
- We update weights iteratively:
  - $w := w - \alpha \frac{\partial J}{\partial w}$
  - $b := b - \alpha \frac{\partial J}{\partial b}$
  - $\alpha$ = learning rate
- Derivatives:
  - $\frac{\partial J}{\partial w} = \frac{1}{m} \sum (\hat{y}^{(i)} - y^{(i)}) x^{(i)}$
  - $\frac{\partial J}{\partial b} = \frac{1}{m} \sum (\hat{y}^{(i)} - y^{(i)})$

## 7. Example (Step by Step)
| Hours studied (x) | Pass/Fail (y) |
|------------------|---------------|
| 1                | 0             |
| 2                | 0             |
| 3                | 0             |
| 4                | 1             |
| 5                | 1             |
| 6                | 1             |

- Linear regression might predict values >1 or <0.
- Logistic regression fits an S-shaped (sigmoid) curve.
- Example: $P(\text{pass}|x=4.5) = 0.8$ → classify as "Pass".

## 8. Graphs (Visual Idea)
- **Sigmoid curve:** S-shaped, goes from 0 to 1.
- **Decision boundary:** At probability = 0.5.
- (Imagine: x-axis = hours studied, y-axis = probability of passing. The curve climbs up, crossing 0.5 near 4 hours.)

## 9. Real-life Example
- **Email spam detection:** 1 = spam, 0 = not spam.
- **Features:** Contains word "win", has attachments, length, etc.
- Logistic regression outputs probability = 0.85 → classify as spam.

## ✅ Summary
- Logistic regression = linear model + sigmoid function + log loss + gradient descent.
