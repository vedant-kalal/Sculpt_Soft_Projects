# Polynomial Regression: Intuitive Notes

## 1. What is Polynomial Regression?
- Extension of Linear Regression.
- **Linear Regression:** fits a straight line:
  $$y = \beta_0 + \beta_1 x + \epsilon$$
- **Polynomial Regression:** fits a curve:
  $$y = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + \cdots + \beta_d x^d + \epsilon$$
  - $x^2, x^3, \ldots, x^d$ are polynomial features.
  - $d$ = degree of the polynomial.
- Still linear in parameters ($\beta$), so it's part of linear models.

## 2. Why Polynomial?
- Some relationships are non-linear.
- **Examples:**
  - House price vs size: Price grows but slows down → curve.
  - Physics (projectile motion): trajectory is parabolic → quadratic relation.

## 3. The Math Behind It
- Training data: $\{(x_1, y_1), (x_2, y_2), \ldots, (x_n, y_n)\}$
- Build a design matrix $X$ with polynomial terms.
- For degree 2:
  $$X = \begin{bmatrix} 1 & x_1 & x_1^2 \\ 1 & x_2 & x_2^2 \\ \vdots & \vdots & \vdots \\ 1 & x_n & x_n^2 \end{bmatrix}$$
- Model: $y = X\beta + \epsilon$
- $\beta = [\beta_0, \beta_1, \beta_2]^T$
- **Solution (Least Squares):**
  $$\hat{\beta} = (X^T X)^{-1} X^T y$$
- It's just Linear Regression on transformed features!

## 4. Example
**Data:**
| x | y |
|---|---|
| 1 | 2 |
| 2 | 6 |
| 3 | 14 |
| 4 | 28 |

- Looks like quadratic.
- Model: $y = \beta_0 + \beta_1 x + \beta_2 x^2$
- Matrix $X$:
  $$X = \begin{bmatrix} 1 & 1 & 1 \\ 1 & 2 & 4 \\ 1 & 3 & 9 \\ 1 & 4 & 16 \end{bmatrix}$$
- Solve with least squares → get coefficients (say $\beta_0=0$, $\beta_1=1$, $\beta_2=1$).
- Final equation: $y = x + x^2$
- Check: For $x=3$, $y=3+9=12$ (close to 14 observed).
- Curve fits better than straight line.

## 5. Visualization Idea
- **Linear Regression:** straight line.
- **Polynomial Regression (degree 2 or 3):** smooth curve hugging the points.

## 6. Linear Example (Straight Relationship)
**Scenario:** Taxi fare vs. distance traveled.
- Formula: $\text{Fare} = 50 + 10 \times \text{Distance (km)}$
- Base fare = ₹50, Each km adds ₹10
| Distance (km) | Fare (₹) |
|--------------|----------|
| 1            | 60       |
| 2            | 70       |
| 3            | 80       |
| 4            | 90       |
- Always grows at a constant rate (straight line).

## 7. Quadratic Example (Parabola)
**Scenario:** Projectile motion (ball thrown upward).
- Height vs. time is quadratic:
  $$h(t) = v_0 t - \frac{1}{2} g t^2$$
- Data (ball thrown upward at 20 m/s, $g \approx 10$ m/s²):
| Time (s) | Height (m) |
|---------|------------|
| 0       | 0          |
| 1       | 15         |
| 2       | 20         |
| 3       | 15         |
| 4       | 0          |
- Goes up, then comes down → U-shaped parabola.

## 8. Cubic Example (S-shape)
**Scenario:** Supply-demand curve in economics (or population growth with carrying capacity).
- Sometimes approximated by cubic:
  $$y = a x^3 + b x^2 + c x + d$$
- Example data (toy population growth model):
| Year | Population (k) |
|------|---------------|
| 1    | 50            |
| 2    | 120           |
| 3    | 300           |
| 4    | 280           |
| 5    | 500           |
- Growth → slows → then accelerates again = cubic-like curve (two bends).

---
## 9. Quick Summary
- **Polynomial Regression = Linear Regression on polynomial features.**
- Equation: $y = \beta_0 + \beta_1 x + \cdots + \beta_d x^d$
- Use when data looks curved, not straight.
- **Caution:** High degree → overfitting.
