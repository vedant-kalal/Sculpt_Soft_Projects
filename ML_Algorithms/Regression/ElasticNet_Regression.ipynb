# 🌟 Elastic Net Regression (Easy + Deep)

## 1. What it is:

Elastic Net is a type of regularized regression.
It combines:

- **Lasso (L1 penalty)** → makes some coefficients exactly 0 → feature selection.
- **Ridge (L2 penalty)** → shrinks coefficients smoothly → keeps correlated features together.

👉 **Elastic Net = best of both worlds.**

## 2. Formula (objective function)
`Loss = MSE + α * (l1_ratio * ∑|wᵢ| + (1 - l1_ratio) * ∑wᵢ²)`

- **MSE** = error between predicted and actual values
- **α** = how strong the penalty is
- **l1_ratio** = how much weight you give to Lasso vs Ridge
  - If `l1_ratio = 1` → pure Lasso
  - If `l1_ratio = 0` → pure Ridge
  - If `l1_ratio = 0.5` → mix of both

## 3. Why use it?

- **Lasso alone:** can randomly drop one feature if two are correlated → unstable
- **Ridge alone:** keeps both but doesn’t remove useless features

**Elastic Net:**
- Keeps groups of correlated features
- Can still zero out irrelevant ones
- Works better when number of features > number of samples (high-dimensional data)

## 4. Example (real-life style)

Suppose you’re predicting house prices with features:
- `size (sqft)`
- `bedrooms`
- `bathrooms`
- `location score`

👉 **Problem:** `size`, `bedrooms`, and `bathrooms` are correlated.

- **Lasso:** might pick only `size` and drop the others (unstable if data changes).
- **Ridge:** will keep all three with small coefficients.
- **Elastic Net:** will keep all three but may shrink them sensibly, and it can still drop irrelevant ones like `location score` if it adds no value.

## 5. Quick Recap

- **Ridge** = shrink (L2)
- **Lasso** = select (L1)
- **Elastic Net** = shrink + select (L1 + L2 combo)

Best when features are many and correlated.
