# Entropy vs Gini in Decision Trees

## 🔹 Entropy
- **Formula:**
  $$
  Entropy(S) = -\sum p_i \log_2(p_i)
  $$
- Measures uncertainty / disorder.
- Slower to compute (needs log).
- More "theoretical" — comes from information theory.

**Use when:**
- You want more balanced splits.
- Dataset has classes with similar probabilities (like 50-50).

**Example:**
- If a class split is 50-50 → Entropy = 1 (maximum uncertainty).
- If split is 90-10 → Entropy ≈ 0.47 (still fairly high impurity).

## 🔹 Gini Index
- **Formula:**
  $$
  Gini(S) = 1 - \sum p_i^2
  $$
- Measures impurity (chance of misclassifying a sample).
- Faster to compute (no logs).
- Used by default in CART (Classification and Regression Trees).

**Use when:**
- You want speed (Gini is computationally cheaper).
- Dataset is large and performance matters.

**Example:**
- If class split is 50-50 → Gini = 0.5 (max impurity).
- If 90-10 → Gini ≈ 0.18 (much smaller than entropy).
- Gini is more sensitive to imbalanced classes.

## 🔹 Example Side by Side
Suppose we have a node with:
- 9 positive samples, 1 negative sample (90-10 split).

**Entropy:**
$$
Entropy = -[0.9 \log_2 0.9 + 0.1 \log_2 0.1] \approx 0.47
$$

**Gini:**
$$
Gini = 1 - (0.9^2 + 0.1^2) = 0.18
$$

- Both agree it’s a “good split” (low impurity), but:
  - Entropy still says “there’s some uncertainty (0.47)”
  - Gini says “this is very pure (0.18)”

## ✅ Rule of Thumb
- **Entropy:** When you care about information theory explanation (text classification, NLP).
- **Gini:** When you care about speed and simplicity (big datasets, standard ML tasks).
- **In practice:** Results are very similar, so most libraries (like scikit-learn) use Gini by default.
