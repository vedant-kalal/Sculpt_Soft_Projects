# K-Nearest Neighbors (KNN): Intuitive Notes

## 1. What is KNN?
- **KNN** is a supervised learning algorithm used mainly for classification, sometimes for regression.
- It doesn't learn weights or equations; instead, it memorizes the training data.
- When a new point arrives, KNN looks at the k closest points (neighbors) and decides its label based on them.
- **Lazy learner:** No model is trained upfront; all computation happens at prediction time.

## 2. How Does It Work? (Steps)
1. **Choose a value of $k$** (e.g., 3, 5, etc.).
2. **Compute the distance** between the new point and all training points.
   - Most common: **Euclidean Distance**:
     $d(p, q) = \sqrt{\sum_{i=1}^n (p_i - q_i)^2}$
3. **Pick the k nearest neighbors** (smallest distances).
4. **Take a vote:**
   - For classification: majority label wins.
   - For regression: average of neighbor values.

## 3. Example (Classification)
| Hours studied | Pass/Fail |
|--------------|-----------|
| 1            | Fail (0)  |
| 2            | Fail (0)  |
| 3            | Fail (0)  |
| 6            | Pass (1)  |
| 7            | Pass (1)  |
| 8            | Pass (1)  |

- New student studies 4 hours. What will KNN predict?
- Compute distance from 4 to each training point.
- Nearest neighbors (e.g., $k=3$): 3, 6, 2.
- Labels: Fail, Pass, Fail. Majority = Fail. So prediction = Fail.

## 4. Graph Intuition
- Imagine a 2D plot (x = hours studied, y = score, color = pass/fail).
- A new point lands somewhere.
- Draw a circle around it to include k neighbors.
- Count which class dominates inside the circle.
- **Decision boundary** in KNN is not a straight line (unlike logistic regression); it's irregular and depends on data distribution.

## 5. Choosing k
- **Small $k$ (like 1):** Very sensitive, can overfit (one noisy point changes prediction).
- **Large $k$:** Smoother decision boundary, but risk of underfitting.
- Usually use odd numbers (3, 5, 7) for binary classification to avoid ties.

## 6. Limitations
- Slow at prediction (must compute distance to all points).
- Memory heavy (stores all training data).
- Struggles with high-dimensional data (curse of dimensionality).

## 7. Real-life Example
- **Movie recommendation:** Finds “nearest” users and recommends movies they liked.
- **Handwriting recognition:** Compares a digit image with stored images to classify.

## ✅ Summary
- KNN is basically: "Tell me who your neighbors are, and I’ll tell you who you are."
