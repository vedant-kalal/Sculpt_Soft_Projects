```python
import os
import pandas as pd
pd.reset_option('display.max_columns')
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sb
%matplotlib inline
import plotly.express as px
```

```python
df = pd.read_csv(path)
```

```python
figure = plt.figure(figsize=(10,6)) 
sb.heatmap(df.isnull())
plt.show()
```

```python

df.columns
```

```python

plt.figure(figsize=(8,5))
sb.histplot(df['Age'], kde=True, color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python

plt.figure(figsize=(6,4))
sb.countplot(x='Heart Attack Risk', data=df, palette='Set2')
plt.title('Heart Attack Risk Count')
plt.xlabel('Heart Attack Risk')
plt.ylabel('Count')
plt.show()
```

```python

plt.figure(figsize=(12,8))
sb.heatmap(df.select_dtypes(include=['float64', 'int64']).corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python

plt.figure(figsize=(8,5))
sb.boxplot(x='Heart Attack Risk', y='Cholesterol', data=df, palette='Set3')
plt.title('Cholesterol by Heart Attack Risk')
plt.xlabel('Heart Attack Risk')
plt.ylabel('Cholesterol')
plt.show()
```

```python

fig = px.scatter(df, x='Age', y='Cholesterol', color='Heart Attack Risk',
                 title='Age vs. Cholesterol by Heart Attack Risk',
                 labels={'Heart Attack Risk': 'Heart Attack Risk'})
fig.show()
```

```python

fig = px.pie(df, names='Heart Attack Risk', title='Heart Attack Risk Proportion',
             color='Heart Attack Risk', color_discrete_sequence=px.colors.sequential.RdBu)
fig.show()
```

```python

ex_hr = df.groupby('Heart Attack Risk')['Exercise Hours Per Week'].mean().reset_index()
fig = px.bar(ex_hr, x='Heart Attack Risk', y='Exercise Hours Per Week',
             title='Average Exercise Hours Per Week by Heart Attack Risk',
             labels={'Heart Attack Risk': 'Heart Attack Risk', 'Exercise Hours Per Week': 'Avg Exercise Hours'})
fig.show()
```

```python

plt.figure(figsize=(8,5))
sb.boxplot(x='Heart Attack Risk', y='Stress Level', data=df, palette='Set2')
plt.title('Stress Level by Heart Attack Risk')
plt.xlabel('Heart Attack Risk')
plt.ylabel('Stress Level')
plt.show()
```

```python

sleep_hr = df.groupby('Heart Attack Risk')['Sleep Hours Per Day'].mean().reset_index()
fig = px.bar(sleep_hr, x='Heart Attack Risk', y='Sleep Hours Per Day',
             title='Average Sleep Hours Per Day by Heart Attack Risk',
             labels={'Heart Attack Risk': 'Heart Attack Risk', 'Sleep Hours Per Day': 'Avg Sleep Hours'})
fig.show()
```

```python

plt.figure(figsize=(6,4))
sb.countplot(x='Sex', hue='Heart Attack Risk', data=df, palette='Set2')
plt.title('Sex vs Heart Attack Risk')
plt.xlabel('Sex')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sb.countplot(x='Diabetes', hue='Heart Attack Risk', data=df, palette='Set1')
plt.title('Diabetes vs Heart Attack Risk')
plt.xlabel('Diabetes')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sb.countplot(x='Family History', hue='Heart Attack Risk', data=df, palette='Set3')
plt.title('Family History vs Heart Attack Risk')
plt.xlabel('Family History')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sb.countplot(x='Smoking', hue='Heart Attack Risk', data=df, palette='Set2')
plt.title('Smoking vs Heart Attack Risk')
plt.xlabel('Smoking')
plt.ylabel('Count')
plt.show()
```
